```python
import os

project_path = "/content/drive/MyDrive/ML Projects/End-to-End-Customer-Churn-ML-Pipeline"

src_path = os.path.join(project_path, "src")

os.makedirs(src_path, exist_ok=True)

files = [
    "model_utils.py"
]

for file in files:
    file_path = os.path.join(src_path, file)
    if not os.path.exists(file_path):
        open(file_path, "w").close()

print("src folder and Python files created successfully!")
```

```python
from src.model_utils import evaluate_model
```

```python
logistic = joblib.load("/content/drive/MyDrive/ML Projects/End-to-End-Customer-Churn-ML-Pipeline/models/logistic_regression.pkl")

tree = joblib.load("/content/drive/MyDrive/ML Projects/End-to-End-Customer-Churn-ML-Pipeline/models/decision_tree.pkl")

forest = joblib.load("/content/drive/MyDrive/ML Projects/End-to-End-Customer-Churn-ML-Pipeline/models/random_forest.pkl")
```

```python
models = {
    "Logistic Regression": logistic,
    "Decision Tree": tree,
    "Random Forest": forest
}
```

```python
X_test = pd.read_csv("/content/drive/MyDrive/ML Projects/End-to-End-Customer-Churn-ML-Pipeline/data/processed/X_test.csv")
y_test = pd.read_csv("/content/drive/MyDrive/ML Projects/End-to-End-Customer-Churn-ML-Pipeline/data/processed/y_test.csv").squeeze()
```

```python
results=[]

for name,model in models.items():

    metrics=evaluate_model(
        model,
        X_test,
        y_test
    )

    metrics["Model"]=name

    results.append(metrics)
```

```python
results = pd.DataFrame(results).sort_values(by="Accuracy",ascending=False)
results
```
